```python
from abc import ABC, abstractmethod

class BaseBook(ABC):
    
    @abstractmethod
    def get_info(self):
        pass
```

```python
class Book(BaseBook):
    def __init__(self, isim, yazar):
        self.__isim = isim
        self.__yazar = yazar
        
    def get_info(self):
        return f"Kitap: {self.__isim}, Yazar: {self.__yazar}"
        
    def __str__(self):
        return self.get_info()
```

```python
class EBook(Book):
    def __init__(self, isim, yazar, boyut):
        super().__init__(isim, yazar)
        self.__boyut = boyut
    def get_info(self):
        return f"{super().get_info()}, Boyut: {self.__boyut} MB"
```

```python
class Library:
    def __init__(self):
        self.books: list[book] = [] 
    def add_book(self, book: Book):
        self.books.append(book)
    def list_books(self):
        for book in self.books:
            print(book)
    def save_to_file(self, filename):
        try:
            with open(filename, mode="w", encoding="utf-8") as newFile:
                for book in self.books:
                    newFile.write(str(book) + "\n") 
            print(f"Başarılı! Kitaplar '{filename}' dosyasına kaydedildi.")
        except Exception as e:
            print(f"Hata oluştu: {e}")
```

```python
my_library = Library()
book1 = Book("Sefiller", "Victor Hugo")
book2 = Book("Şeker Portakalı", "José Mauro de Vasconcelos")
book3 = Book("Dönüşüm", "Franz Kafka")
ebook1 = EBook("100 günde Python", "Anonim", 2)
ebook2 = EBook("Yazılımda Gelişmeler", "Anonim", 1)
my_library.add_book(book1)
my_library.add_book(book2)
my_library.add_book(book3)
my_library.add_book(ebook1)
my_library.add_book(ebook2)
print("~~ KÜTÜPHANEDEKİ KİTAPLAR ~~")
my_library.list_books()
my_library.save_to_file("kutuphane.txt")
```
