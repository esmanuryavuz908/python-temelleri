```python
en_yuksek_skor = 0
isim = input("Lütfen isminizi giriniz!: ")
print(f"Hoş geldin, {isim}! Akıllı Quiz'e başlıyoruz.")

while True:
    print("\n--- ANA MENÜ ---")
    print("1 - Quiz Başlat")
    print("2 - Çıkış")
    
    secim = input("Seçimin (1 veya 2): ")

    if secim == "2":
        print("Oyundan çıkılıyor. Görüşmek üzere!")
        break
        
    elif secim == "1":
        print("\nQuiz Başlıyor! Başarılar...\n")
        
        puan = 0
        yanlis_sayisi = 0
        
        sorular = [
            ("Soru 1: 5 + 3 = ?\na) 6\nb) 8\nc) 9", "b"),
            ("Soru 2: Veri Bilimi için en çok hangi dili kullanırız?\na) Python\nb) HTML\nc) CSS", "a"),
            ("Soru 3: 10 / 2 = ?\na) 5\nb) 4\nc) 6", "a"),
            ("Soru 4: 'She ___ a student.' cümlesindeki boşluğa ne gelmelidir?\na) am\nb) is\nc) are", "b"),
            ("Soru 5: 'Elma' kelimesinin İngilizcesi nedir?\na) Banana\nb) Orange\nc) Apple", "c")
        ]
        
        for soru_metni, dogru_cevap in sorular:
            print(soru_metni)
            cevap = input("Cevabın nedir? (a/b/c veya 'ipucu'): ").lower()
            
            if cevap == "ipucu":
                pass
                continue
                
            if cevap == "":
                print("Uyarı: Boş giriş yaptınız! Soru atlanıyor.")
                continue
                
            if cevap == dogru_cevap:
                print("Doğru!\n")
                puan += 10
            else:
                print("Yanlış!\n")
                yanlis_sayisi += 1
                
                if yanlis_sayisi == 3:
                    print("Çok fazla yanlış yaptın!")
                    break
        
        print(f"\n--- OYUN BİTTİ ---")
        print(f"Toplam Puanın: {puan}, Toplam Yanlışın: {yanlis_sayisi}")
        
        if puan <= 20:
            print("Seviyen: Başlangıç")
        elif 30 <= puan <= 40:
            print("Seviyen: Orta")
        elif puan >= 50:
            print("Seviyen: Uzman")
            
        if puan > en_yuksek_skor:
            en_yuksek_skor = puan
            print(f" YENİ REKOR! En yüksek skor artık: {en_yuksek_skor}")
        else:
            print(f"En yüksek skorun: {en_yuksek_skor}")
            
        tekrar = input("\nTekrar oynamak ister misin? (evet/hayır): ").lower()
        if tekrar != "evet":
            print(f"Görüşmek üzere {isim}!")
            break
            
    else:
        print("Geçersiz seçimde bulundunuz, lütfen 1 veya 2'yi seçin.")
 
```
