```python
def kullanici_ekle():
    isim = input("İsminizi giriniz:")
    yas = int(input("Yasınızı giriniz:"))
    sehir = input("Sehrinizi giriniz:")
    yetenekler = input("Yeteneklerinizi giriniz:")
    yetenek_listesi = yetenekler.split(",")
    kullanici = {"isim":isim,"yas":yas,"sehir":sehir,"yetenekler":yetenek_listesi}
    kullanicilar.append(kullanici)
    print("Kayıt işlemi başarıyla tamamlandı!")
```

```python
def kullanicilari_goster():
    if len(kullanicilar) == 0:
        print("Henüz kayıtlı kullanıcı yok!!!")
    else:
        for kisi in kullanicilar:
            print(f"İsim: {kisi['isim']} | Yaş: {kisi['yas']} | Şehir: {kisi['sehir']} | Yetenekler: {kisi['yetenekler']}")
```

```python
def yas_filtreleme():
    yas_siniri = int(input("Yas siniri giriniz: "))
    for kisi in kullanicilar:
        if kisi["yas"] >= yas_siniri: 
            print(f"İsim: {kisi['isim']} | Yaş: {kisi['yas']} | Şehir: {kisi['sehir']} | Yetenekler: {kisi['yetenekler']}")
```

```python
def sehir_filtreleme():
    girilen_sehir = input("Bir sehir giriniz: ")

    for kisi in kullanicilar:
        if kisi["sehir"] == girilen_sehir:
            print(f"İsim: {kisi['isim']} | Yaş: {kisi['yas']} | Şehir: {kisi['sehir']} | Yetenekler: {kisi['yetenekler']}")
```

```python
def benzersiz_yetenekleri_gosterme():
    tum_yetenekler = list() 
    for kisi in kullanicilar:
        tum_yetenekler.extend(kisi["yetenekler"]) 
    benzersiz_yetenekler = set(tum_yetenekler)
    print(benzersiz_yetenekler)
```

```python
def menu_goster():
    menu = (
    "1- Kullanıcı ekle",
    "2- Kullanıcıları göster",
    "3- Yaş filtresi",
    "4- Şehir filtresi",
    "5- Benzersiz yetenekler",
    "0- Çıkış"
)
```

```python
while True:
    menu_goster()
    secim = input("Lütfen 0 ile 5 arasında bir sayı giriniz: ")
    
    if secim == "1":
        kullanici_ekle()
    elif secim == "2":
        kullanicilari_goster()
    elif secim == "3":
        yas_filtreleme()
    elif secim == "4":
        sehir_filtreleme()
    elif secim == "5":
        benzersiz_yetenekleri_gosterme()
    elif secim == "0":
        print("Programdan çıkılıyor. İyi günler!")
        break
    else:
        print("Hatalı giriş yaptınız, lütfen tekrar deneyin.") 
```
